```python
import yfinance as yf
import pandas as pd

# ── Configuration ────────────────────────────────────────────────────────────
TICKER      = "^GSPC"          # S&P 500 index
START_DATE  = "2008-04-07"     # matches period1=1523059200 in the URL
END_DATE    = "2026-04-08"     # today (or set to None for "up to today")
INTERVAL    = "1d"             # daily data  (options: 1d, 1wk, 1mo)
OUTPUT_CSV  = "sp500_history.csv"
# ─────────────────────────────────────────────────────────────────────────────


def fetch_sp500(ticker: str, start: str, end: str, interval: str) -> pd.DataFrame:
    """Download historical OHLCV data from Yahoo Finance via yfinance."""
    print(f"Downloading {ticker} from {start} to {end} ...")
    df = yf.download(
        tickers=ticker,
        start=start,
        end=end,
        interval=interval,
        auto_adjust=True,   # adjusts for splits/dividends automatically
        progress=True,
    )
    return df


def clean_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """Rename columns and reset the index for a tidy table."""
    df = df.copy()
    df.index.name = "Date"
    df.reset_index(inplace=True)

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [col[0] for col in df.columns]

    float_cols = df.select_dtypes(include="float").columns
    df[float_cols] = df[float_cols].round(2)

    return df


def main():
    df_raw   = fetch_sp500(TICKER, START_DATE, END_DATE, INTERVAL)
    df_clean = clean_dataframe(df_raw)

    print(f"\n✅  Retrieved {len(df_clean):,} rows.\n")
    print(df_clean.head(10).to_string(index=False))
    print("  ...")
    print(df_clean.tail(5).to_string(index=False))

    # ── Save to CSV ──────────────────────────────────────────────────────────
    df_clean.to_csv(OUTPUT_CSV, index=False)
    print(f"\n💾  Saved to '{OUTPUT_CSV}'")

    # ── Basic stats ──────────────────────────────────────────────────────────
    print("\n── Summary statistics ─────────────────────────────────────────")
    print(df_clean[["Open", "High", "Low", "Close", "Volume"]].describe().round(2))

    return df_clean


df = main()
```
